```python
df['HourOfDay'] = (df['Time'] // 3600) % 24

fraud  = df[df['Class'] == 1]['HourOfDay'].value_counts(normalize=True).sort_index()
normal = df[df['Class'] == 0]['HourOfDay'].value_counts(normalize=True).sort_index()

hours = pd.DataFrame({'fraud': fraud, 'normal': normal}).fillna(0)
hours['ratio'] = hours['fraud'] / hours['normal']
hours
```

```python
hours[['fraud', 'normal']].plot(kind='bar', figsize=(12, 4))
```
